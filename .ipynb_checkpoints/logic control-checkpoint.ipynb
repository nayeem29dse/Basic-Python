### Problem 1 —Write a function bmi_category(bmi) that returns:
    "underweight" if BMI < 18.5
    "normal" if 18.5 ≤ BMI < 25
    "overweight" if 25 ≤ BMI < 30
    "obese" if BMI ≥ 30
Test it with values: 17.0, 22.4, 27.1, 35.0

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return "underweight" 
    elif 18.5 <= bmi < 25:
        return "normal"
    elif 25 <= bmi < 30:
        return "overweight" 
    else:
        return "obese" 
print(bmi_category(17.0))
print(bmi_category(22.4))
print(bmi_category(27.1))
print(bmi_category(35.0))
```

### Problem 2 — Truthiness:
    You have a list of exam scores. Write code that prints "valid dataset" if the list is non-empty AND the maximum score doesn't exceed 100. Otherwise print what the specific problem is.
Test with: [], [85, 92, 78], [88, 105, 76]

```python
def check_score(score):
    if  not score:
        return "empty dataset"
    elif max(score)>100:
        return "score exceeds 100"
    else:
        return "valid dataset"
print(check_score([]))
print(check_score([85, 92, 78]))
print(check_score([88, 105, 76]))
```

### Problem 3 — Short circuit + chained logic:
    Write a function can_enroll(age, gpa, has_prereq) that returns True only if:
    age is between 18 and 60 (inclusive), AND gpa ≥ 2.5, AND has_prereq is True
But here's the constraint: if gpa is None, the function must not crash — it should return False.

```python
def can_enroll(age, gpa, has_prereq):
    if gpa is None:
        return False
    elif 18<=age<=60 and gpa>=2.5 and has_prereq:
        return True
    else:
        return False
print(can_enroll(18,None,True))
```

### Problem 4 — Applied (statistical flavour):
    Write a function interpret_zscore(z) that returns:
    "extreme outlier" if |z| > 3
    "mild outlier" if 2 < |z| ≤ 3
    "within normal range" otherwise
Use the absolute value (abs()) and a ternary expression somewhere in your solution.

```python
def interpret_zscore(z):
    if abs(z) > 3:
        return "extreme outlier" 
    elif 2 < abs(z) <= 3:
        return "mild outlier"
    else:
        return "within normal range"
print(interpret_zscore(2))
```
