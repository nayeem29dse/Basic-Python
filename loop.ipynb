# Loop

## Before syntax, the mindset.
    A loop is not just "repeat something." It is traversing a structure to compute a result. Every loop you write in data science is doing one of four things:
    
    Accumulating — building up a result (sum, count, list)
    Filtering — keeping only what meets a condition
    Transforming — changing each element into something else
    Searching — finding the first element that satisfies a condition
    
    Before writing any loop, ask yourself: which of these four am I doing? That question will later tell you which tool to use — a for loop, a comprehension, or a NumPy vectorized operation.

### Concept 1: for loop

```python
scores = [88, 72, 95, 61, 83]

# Accumulating
total = 0
for score in scores:
    total += score
print(total / len(scores))   # mean: 79.8

# Transforming
scaled = []
for score in scores:
    scaled.append(score / 100)
print(scaled)   # [0.88, 0.72, 0.95, 0.61, 0.83]
```

### The loop variable (score) is just a name. It takes the value of each element one at a time. The collection is never modified — you build something new.

## Concept 2: enumerate and zip
These two are essential in data work. You will use them daily.

```python
# enumerate — when you need both index and value
features = ["age", "bmi", "glucose", "insulin"]

for i, feature in enumerate(features):
    print(f"{i}: {feature}")
# 0: age
# 1: bmi
# 2: glucose
# 3: insulin

# zip — when you need to walk two lists together
actual    = [1, 0, 1, 1, 0]
predicted = [1, 0, 0, 1, 0]

correct = 0
for a, p in zip(actual, predicted):
    if a == p:
        correct += 1
print(correct / len(actual))   # accuracy: 0.8
```

### zip stops at the shorter list. If your lists are different lengths, you will silently lose data. Always verify lengths match before zipping.

## Concept 3: while loop
Use while when you don't know in advance how many iterations you need.

```python
# Gradient descent stopping condition
error = 100.0
learning_rate = 0.1
iteration = 0

while error > 0.01:
    error = error * (1 - learning_rate)
    iteration += 1

print(f"Converged in {iteration} iterations")
```

### Always ask: can this loop run forever? Every while loop needs a guaranteed exit condition. In the example above, error shrinks each iteration — it will always reach 0.01 eventually. If your condition can never become False, you have an infinite loop.

## Concept 4: break and continue

```python
# break — exit the loop immediately
values = [4, 7, 2, -1, 9, 3]
for v in values:
    if v<0:
        print("negative value found, stopping")
        break
print(v)
```

```python
# continue — skip this iteration, move to next
values = [4, 7, 2, -1, 9, 3]
for v in values:
    if v<0:
        print("negative value found")
        continue
    print(v)
```

## Concept 5: List comprehensions
A comprehension is a compact loop that always produces a new list. It maps directly onto the transform and filter patterns.

```python
scores = [88, 72, 95, 61, 83]

# Transform — no condition
scaled = [s / 100 for s in scores]

# Filter — with condition
passing = [s for s in scores if s >= 75]

# Transform + filter — both together
scaled_passing = [s / 100 for s in scores if s >= 75]
```

### The structure is always:
    python
    [expression for item in iterable if condition]

### The if condition part is optional.
Dict comprehension — you will use this constantly in data work:

```python
features = ["age", "bmi", "glucose"]
means    = [45.2, 27.3, 112.8]

feature_means = {f: m for f, m in zip(features, means)}
# {"age": 45.2, "bmi": 27.3, "glucose": 112.8}
```

## Concept 6: Nested loops

```python
# Confusion matrix — 2D iteration
matrix = [[50, 5],
          [8,  37]]
for i,row in enumerate(matrix):
    for j, val in enumerate(row):
        print(f"row {i}, col {j}: {val}")
```

### In data science, nested loops over large datasets are slow. When you see yourself writing a nested loop over data, that's a signal that NumPy vectorization might be the right tool.

    Test with [4, 8, 15, 16, 23, 42].
    Variance formula: mean of squared deviations — (1/n) * Σ(xi - mean)²

```python
def descriptive_stats(data):
    total=0
    n=0
    for value in data:
        total+=value
        n+=1
#For mean
    mean=total/n
#For varience
    smallest = data[0]
    largest = data[0]
    squared_diff_sum = 0
    for value in data:
        diff=value-mean
        squared_diff_sum+=diff**2
    variance=squared_diff_sum/n
#For range
    for value in data:
        if value < smallest:
            smallest = value

        if value > largest:
            largest = value
    range=largest-smallest
    return {
        "mean": mean,
        "variance": variance,
        "range": range
    }       
print(descriptive_stats([4, 8, 15, 16, 23, 42]))
```

### Problem 2 — Transform + filter:
    Write a function normalize_and_filter(data, threshold) that:
    
    Min-max normalizes each value: (x - min) / (max - min)
    Returns only the normalized values that are above threshold
    Returns an empty list if all values are identical (would cause division by zero)
    
    Use a comprehension for the normalization step and another for the filter step. Do not use min() or max() — compute them with loops first.
    Test with [10, 20, 30, 40, 50] and threshold 0.5.

```python
def normalize_and_filter(data, threshold):

    smallest = data[0]
    largest = data[0]
    for value in data:
        if value < smallest:
            smallest = value

        if value > largest:
            largest = value
    normalized = [(x - smallest) / (largest - smallest) for x in data]
    normalized_data = [value for value in normalized if value > threshold]
    return normalized_data
print(normalize_and_filter([10, 20, 30, 40, 50], 0.5))
```

## Problem 3 — zip + enumerate:
    You have two lists — actual and predicted — of equal length containing class labels 0 or 1.
    Write a function classification_report(actual, predicted) that returns a dictionary with:
    "accuracy" — proportion of correct predictions
    "true_positives" — predicted 1, actual 1
    "true_negatives" — predicted 0, actual 0
    "false_positives" — predicted 1, actual 0
    "false_negatives" — predicted 0, actual 1    
    Use zip. Do not use any library functions.
    Test with:
    actual    = [1, 0, 1, 1, 0, 1, 0, 0]
    predicted = [1, 0, 0, 1, 0, 1, 1, 0]

```python
import pprint
def classification_report(actual, predicted):
    true_positives=0
    true_negatives=0
    false_positives=0
    false_negatives=0
    for a, p in zip(actual, predicted):
        if a==1 and p==1:
            true_positives+=1
        elif a==0 and p==0:
            true_negatives+=1
        elif a==0 and p==1:
            false_positives+=1
        elif a==1 and p==0:
            false_negatives+=1
    correct=true_positives+true_negatives
    accuracy=correct/len(predicted)
    return {
        "accuracy": accuracy,
        "true_positives": true_positives,
        "true_negatives": true_negatives,
        "false_positives": false_positives,
        "false_negatives": false_negatives
    }
pprint.pprint(classification_report([1, 0, 1, 1, 0, 1, 0, 0],[1, 0, 0, 1, 0, 1, 1, 0]))
            
```

## Problem 4 — while + real logic:
    Write a function simulate_learning_curve(initial_error, learning_rate, max_iterations) that simulates gradient descent:
    Each iteration: error = error * (1 - learning_rate)
    Stop when error < 0.001 OR when max_iterations is reached
    Return a dictionary with "final_error", "iterations", and "converged" (True if stopped due to error threshold, False if hit max iterations)
    Test with:
    simulate_learning_curve(1.0, 0.3, 100)   # should converge
    simulate_learning_curve(1.0, 0.01, 10)   # should not converge

```python
def simulate_learning_curve(initial_error, learning_rate, max_iterations):
    error = initial_error
    iterations = 0
    # Run gradient descent simulation
    while iterations < max_iterations:

        error = error * (1 - learning_rate)
        iterations += 1

        # Stop if error becomes very small
        if error < 0.001:
            return {
                "final_error": error,
                "iterations": iterations,
                "converged": True
            }
    # If max iterations reached
    return {
        "final_error": error,
        "iterations": iterations,
        "converged": False
    }
# Test 1: Should converge
result1 = simulate_learning_curve(1.0, 0.3, 100)
print(result1)
# Test 2: Should NOT converge
result2 = simulate_learning_curve(1.0, 0.01, 10)
print(result2)
    
    
```

    Write a function running_statistics(data) that iterates through the list once and returns a dictionary with "mean", "min", "max", and "count". One loop only — no separate loops for each statistic.
    Test with [33, 7, 52, 18, 44, 29, 61, 5].

```python
def running_statistics(data):
    sum=0
    n=0
    for value in data:
        sum+=value
        n+=1
    avg= sum/n
    smallest=data[0]
    largest=data[0]
    for value in data:
        if value< smallest:
            smallest=value
        if value > largest:
            largest=value
    return {"mean":avg,
            "min": smallest,
            "max": largest,
           "count":n}
print(running_statistics([33, 7, 52, 18, 44, 29, 61, 5]))
```
