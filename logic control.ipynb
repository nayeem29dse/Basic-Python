### Problem 1 —Write a function bmi_category(bmi) that returns:
    "underweight" if BMI < 18.5
    "normal" if 18.5 ≤ BMI < 25
    "overweight" if 25 ≤ BMI < 30
    "obese" if BMI ≥ 30
Test it with values: 17.0, 22.4, 27.1, 35.0

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return "underweight" 
    elif 18.5 <= bmi < 25:
        return "normal"
    elif 25 <= bmi < 30:
        return "overweight" 
    else:
        return "obese" 
print(bmi_category(17.0))
print(bmi_category(22.4))
print(bmi_category(27.1))
print(bmi_category(35.0))
```

### Problem 4 — Applied (statistical flavour):
    Write a function interpret_zscore(z) that returns:
    "extreme outlier" if |z| > 3
    "mild outlier" if 2 < |z| ≤ 3
    "within normal range" otherwise
Use the absolute value (abs()) and a ternary expression somewhere in your solution.

```python
def interpret_zscore(z):
    if abs(z) > 3:
        return "extreme outlier" 
    elif 2 < abs(z) <= 3:
        return "mild outlier"
    else:
        return "within normal range"
print(interpret_zscore(2))
```

### You have a list of model accuracy scores. Write a function validate_accuracies(scores) that:
    Returns "no scores provided" if the list is empty
    Returns "invalid: accuracies must be between 0 and 1" if any score is outside [0, 1]
    Returns "valid" otherwise
Do not use min()/max() — use a loop or any().

```python
def validate_accuracies(scores):
    if not scores:
        return "no scores provided"
    if any(score < 0 or score > 1 for score in scores):
        return "invalid: accuracies must be between 0 and 1"
    return "valid"

print(validate_accuracies([]))
print(validate_accuracies([0.85, 0.92, 0.78]))
print(validate_accuracies([0.88, 1.05, 0.76]))
print(validate_accuracies([-0.1, 0.9]))
```

### Write a function classify_sample(age, bmi, glucose) that returns a diabetes risk level:
    "high risk" if glucose > 140 OR (bmi > 30 AND age > 45)
    "moderate risk" if glucose > 100 OR bmi > 25
    "low risk" otherwise
The catch: any of the three inputs can be None. If a value is None, treat that condition as False — the function must never crash regardless of input.
Test it with (None, None, None), (50, 32, 130), (30, 22, 150).

```python
def classify_sample(age, bmi, glucose):
    high_risk = (
        (glucose is not None and glucose > 140) or
        (
            bmi is not None and bmi > 30 and
            age is not None and age > 45
        )
    )

    moderate_risk = (
        (glucose is not None and glucose > 100) or
        (bmi is not None and bmi > 25)
    )

    return (
        "high risk" if high_risk else
        "moderate risk" if moderate_risk else
        "low risk"
    )
print(classify_sample(None, None, None))
print(classify_sample(50, 32, 130))
print(classify_sample(30, 22, 150))
```

### Chained conditions with data validation:
    Write a function validate_dataset(n_samples, n_features, target_ratio) where:
    n_samples is an integer
    n_features is an integer
    target_ratio is a float representing the ratio of positive class (e.g. 0.3 means 30% positive)
    Return:

    "invalid: insufficient samples" if n_samples < 30
    "invalid: no features" if n_features < 1
    "invalid: target ratio out of range" if target_ratio is not in [0, 1]
    "warning: severe class imbalance" if target_ratio < 0.1 or target_ratio > 0.9
    "valid" otherwise
Any input can be None — handle it without crashing. Order of checks matters.

```python
def  validate_dataset(n_samples, n_features, target_ratio):
    if n_samples is None or n_samples<30:
        return "invalid: insufficient samples"
    elif n_features is None or n_features<1:
        return "invalid: no features"
    elif target_ratio is None or (target_ratio<0 or target_ratio>1):
       return "invalid: target ratio out of range"
    elif target_ratio < 0.1 or target_ratio > 0.9:
        return "warning: severe class imbalance"
    else:
        return "valid"
print(validate_dataset(30,1,0.1))
print(validate_dataset(0,1,0.1))
print(validate_dataset(0,1,0)) 
print(validate_dataset(30, 1, 1.5))
print(validate_dataset(30, 1, None))
        
```

### Boolean logic in data cleaning:
    Write a function should_impute(value, col_type, missing_threshold) that returns True if a missing value should be imputed, based on these rules:
    
    value must be None (if it's not missing, never impute)
    col_type must be either "numeric" or "categorical"
    missing_threshold must be below 0.4 (if more than 40% of the column is missing, don't impute — drop instead)
    
    Return False in all other cases including if any argument is None.
    Test with at least 5 combinations.

```python
def should_impute(value, col_type, missing_threshold):
    if value is  None:
        if col_type in("numeric", "categorical"):
            if missing_threshold is not None and missing_threshold<0.4:
                return True
    return False
print(should_impute(None, "numeric", 0.3))
print(should_impute(None, "categorical", 0.2))
print(should_impute(None, "numeric", 0.5))
print(should_impute(25, "numeric", 0.3))
print(should_impute(None, "text", 0.2))
print(should_impute(None, None, 0.2))
```

## any() and all():
    You have not seen all() yet — figure out what it does from context and use it.
    Write a function is_valid_probability_distribution(probs) that returns True only if:
    
    probs is non-empty
    all values are between 0 and 1 inclusive
    the values sum to approximately 1.0 (within 0.01 tolerance — use abs())
    
    Test with:
    python[0.2, 0.3, 0.5]       # True
    [0.2, 0.3, 0.4]       # False — doesn't sum to 1
    [0.2, -0.3, 1.1]      # False — values out of range
    []                     # False — empty

```python
def is_valid_probability_distribution(probs):
    if probs:
        if all (0<=p<=1 for p in probs):
            total=sum(probs)
            if abs(total-1.0)<=0.01:
                return True
    return False
print(is_valid_probability_distribution([0.2, 0.3, 0.5]))
print(is_valid_probability_distribution([0.2, 0.3, 0.4]))
print(is_valid_probability_distribution([0.2, -0.3, 1.1]))
print(is_valid_probability_distribution([]))

    
```

## Hardest, applied:
    Write a function model_selection_advice(accuracy, precision, recall, use_case) where use_case is either "medical" or "spam".
    Rules:
    
    If any of accuracy, precision, recall is None, return "insufficient metrics"
    For "medical" use case: recall must be ≥ 0.9 (missing a disease is worse than a false alarm). If recall ≥ 0.9 and accuracy ≥ 0.8, return "recommended". If recall ≥ 0.9 but accuracy < 0.8, return "acceptable". Otherwise return "not recommended"
    For "spam" use case: precision must be ≥ 0.95 (false positives delete real emails). If precision ≥ 0.95 and accuracy ≥ 0.85, return "recommended". If precision ≥ 0.95 but accuracy < 0.85, return "acceptable". Otherwise return "not recommended"
    If use_case is anything else, return "unknown use case"
    
    Test with at least one case per outcome.

```python
def model_selection_advice(accuracy, precision, recall, use_case):
    if accuracy is None or precision is None or recall is None:
        return "insufficient metrics"

    if use_case == "medical":

        if recall >= 0.9:

            if accuracy >= 0.8:
                return "recommended"
            else:
                return "acceptable"

        return "not recommended"
    elif use_case == "spam":

        if precision >= 0.95:

            if accuracy >= 0.85:
                return "recommended"
            else:
                return "acceptable"

        return "not recommended"
    return "unknown use case"
print(model_selection_advice(0.85, 0.70, 0.92, "medical"))
print(model_selection_advice(0.75, 0.80, 0.95, "medical"))
print(model_selection_advice(0.90, 0.90, 0.70, "medical"))
print(model_selection_advice(0.90, 0.97, 0.80, "spam"))
print(model_selection_advice(0.80, 0.96, 0.75, "spam"))
print(model_selection_advice(0.92, 0.80, 0.90, "spam"))
print(model_selection_advice(None, 0.95, 0.90, "medical"))
print(model_selection_advice(0.9, 0.9, 0.9, "finance"))
   
```
