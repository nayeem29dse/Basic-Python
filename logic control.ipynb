### Problem 1 —Write a function bmi_category(bmi) that returns:
    "underweight" if BMI < 18.5
    "normal" if 18.5 ≤ BMI < 25
    "overweight" if 25 ≤ BMI < 30
    "obese" if BMI ≥ 30
Test it with values: 17.0, 22.4, 27.1, 35.0

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return "underweight" 
    elif 18.5 <= bmi < 25:
        return "normal"
    elif 25 <= bmi < 30:
        return "overweight" 
    else:
        return "obese" 
print(bmi_category(17.0))
print(bmi_category(22.4))
print(bmi_category(27.1))
print(bmi_category(35.0))
```

### Problem 2 — Truthiness:
    You have a list of exam scores. Write code that prints "valid dataset" if the list is non-empty AND the maximum score doesn't exceed 100. Otherwise print what the specific problem is.
Test with: [], [85, 92, 78], [88, 105, 76]

```python
def check_score(score):
    if  not score:
        return "empty dataset"
    elif max(score)>100:
        return "score exceeds 100"
    else:
        return "valid dataset"
print(check_score([]))
print(check_score([85, 92, 78]))
print(check_score([88, 105, 76]))
```

### Problem 3 — Short circuit + chained logic:
    Write a function can_enroll(age, gpa, has_prereq) that returns True only if:
    age is between 18 and 60 (inclusive), AND gpa ≥ 2.5, AND has_prereq is True
But here's the constraint: if gpa is None, the function must not crash — it should return False.

```python
def can_enroll(age, gpa, has_prereq):
    if gpa is None:
        return False
    elif 18<=age<=60 and gpa>=2.5 and has_prereq:
        return True
    else:
        return False
print(can_enroll(18,None,True))
```

### Problem 4 — Applied (statistical flavour):
    Write a function interpret_zscore(z) that returns:
    "extreme outlier" if |z| > 3
    "mild outlier" if 2 < |z| ≤ 3
    "within normal range" otherwise
Use the absolute value (abs()) and a ternary expression somewhere in your solution.

```python
def interpret_zscore(z):
    if abs(z) > 3:
        return "extreme outlier" 
    elif 2 < abs(z) <= 3:
        return "mild outlier"
    else:
        return "within normal range"
print(interpret_zscore(2))
```

### Write a function loan_eligibility(income, credit_score, has_default) that returns:
    "eligible" if income ≥ 50000 AND credit_score ≥ 700 AND has_default is False
    "review required" if income ≥ 50000 AND credit_score ≥ 600 AND has_default is False
    "ineligible" in all other cases
If credit_score is None, return "ineligible" without crashing.
Test with at least 4 combinations that cover all three outcomes.

```python
def loan_eligibility(income, credit_score, has_default):
    if credit_score is None:
        return "ineligible"
    elif income >= 50000 and credit_score >= 700 and not has_default:
        return "eligible"
    elif income >= 50000 and credit_score >= 600 and not has_default:
        return "review required"
    else:
        return "ineligible"
print(loan_eligibility(100000, 750, False))
print(loan_eligibility(100000, None, False))
print(loan_eligibility(50000, 500, False))
print(loan_eligibility(10000, 250, True))
```

### You have a list of model accuracy scores. Write a function validate_accuracies(scores) that:
    Returns "no scores provided" if the list is empty
    Returns "invalid: accuracies must be between 0 and 1" if any score is outside [0, 1]
    Returns "valid" otherwise
Do not use min()/max() — use a loop or any().

```python
def validate_accuracies(scores):
    if not scores:
        return "no scores provided"
    if any(score < 0 or score > 1 for score in scores):
        return "invalid: accuracies must be between 0 and 1"
    return "valid"

print(validate_accuracies([]))
print(validate_accuracies([0.85, 0.92, 0.78]))
print(validate_accuracies([0.88, 1.05, 0.76]))
print(validate_accuracies([-0.1, 0.9]))
```

### Write a function classify_sample(age, bmi, glucose) that returns a diabetes risk level:
    "high risk" if glucose > 140 OR (bmi > 30 AND age > 45)
    "moderate risk" if glucose > 100 OR bmi > 25
    "low risk" otherwise
The catch: any of the three inputs can be None. If a value is None, treat that condition as False — the function must never crash regardless of input.
Test it with (None, None, None), (50, 32, 130), (30, 22, 150).

```python
def classify_sample(age, bmi, glucose):
    high_risk = (
        (glucose is not None and glucose > 140) or
        (
            bmi is not None and bmi > 30 and
            age is not None and age > 45
        )
    )

    moderate_risk = (
        (glucose is not None and glucose > 100) or
        (bmi is not None and bmi > 25)
    )

    return (
        "high risk" if high_risk else
        "moderate risk" if moderate_risk else
        "low risk"
    )
print(classify_sample(None, None, None))
print(classify_sample(50, 32, 130))
print(classify_sample(30, 22, 150))
```
